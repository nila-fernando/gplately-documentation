# Loading in Your Own Plate Model

This tutorial shows how to load in your own plate model. The following works with your local plate model directory, assuming it follows the expected structure (i.e., .rot for rotations, .gpml for topologies, and a .shp file for static polygons). 

### Imports

Let us set up our packages:

```python
# Ensure you have the required packages installed:
# Run `pip install gplately pygplates` if not already installed.
import gplately
from gplately import pygplates
```

### Loading in a Plate Model

Let's use local files to create a plate reconstruction model based on the Zahirovic et al. (2022) global plate model, [“Subduction and carbonate platform interactions”](https://rmets.onlinelibrary.wiley.com/doi/10.1002/gdj3.146).

In the code below, we define the full file paths for the rotation and topology files manually. These files contain the information needed to reconstruct how tectonic plates have moved and deformed through time.

If you're working with many files or different folders, you can optionally use Python's `os` and `glob` libraries to automatically find files instead of listing them one by one!

```python
# Define the directory containing the plate model files
plate_model_dir = '../../data/plate-model-repo/zahirovic2022'

# Rotation files (Explicitly listed for clarity)
rotation_filenames = [
    f'{plate_model_dir}/Rotations/CombinedRotations.rot',
]

# Topology files (Explicitly listed for clarity)
topology_filenames = [
    f'{plate_model_dir}/Topologies/Deforming_Networks_Active.gpml',
    f'{plate_model_dir}/Topologies/Deforming_Networks_Inactive.gpml',
    f'{plate_model_dir}/Topologies/Feature_Geometries.gpml',
    f'{plate_model_dir}/Topologies/Plate_Boundaries.gpml',
]

# Static polygon file (Explicitly listed for clarity)
static_polygon_filename = f'{plate_model_dir}/StaticPolygons/Global_EarthByte_GPlates_PresentDay_StaticPlatePolygons.shp'
```

To generate the plate reconstruction model, you will need three types of files:

- Rotation files: usually ending with .rot

- Topology feature files: usually ending with .gpml or .gpmlz

- Static polygon files

These files contain different geological data and need to be loaded into specific pyGPlates objects:

- Rotation files → load into a pygplates.RotationModel object

- Topology feature files → load into a pygplates.FeatureCollection object

- Static polygon files → also load into a pygplates.FeatureCollection object

After collecting the file paths, you will use them to create these pyGPlates objects, which are essential for reconstructing plate motions and geological features through time.

```python
# Create RotationModel object from the rotation files
rotation_model_2022 = pygplates.RotationModel(rotation_filenames)

# Create FeatureCollection for topology features
topology_features_2022 = pygplates.FeatureCollection()
for topology_filename in topology_filenames:
    # Omit files with the string "Inactive" in the filepath
    if "Inactive" not in topology_filename:
        topology_features_2022.add(pygplates.FeatureCollection(topology_filename))

# Create FeatureCollection for static polygons
static_polygons_2022 = pygplates.FeatureCollection(static_polygon_filename)
```

Once we have our rotation model, topology features and static polygons, we can supply them to the `PlateReconstruction` object to construct the plate motion model. The `PlateReconstruction` object contains methods to reconstruct topology features to a specific geological time!

You can explore the full syntax and use-cases in the [GPlately documentation for PlateReconstruction](https://gplates.github.io/gplately/latest/sphinx/html/generated/gplately.PlateReconstruction.html#gplately.PlateReconstruction).

```python
# Create a PlateReconstruction model with the following inputs:
# - rotation_model: required; defines the rotation files
# - topology_features: optional; defines the topological features files
# - static_polygons: optional; defines the static polygons

model = gplately.PlateReconstruction(rotation_model_2022, topology_features_2022, static_polygons_2022)
```

Now that you have your plate motion model, you can reconstruct and plot ancient coastlines, plate boundaries, subduction zones, mid-ocean ridges, deforming regions, fossil locations, and more; unlocking a dynamic view of Earth's geological past!
