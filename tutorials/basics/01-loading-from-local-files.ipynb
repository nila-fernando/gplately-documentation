To generate the plate reconstruction model, you will need three types of files:

- Rotation files: usually ending with .rot

- Topology feature files: usually ending with .gpml or .gpmlz

- Static polygon files

These files contain different geological data and need to be loaded into specific pyGPlates objects:

- Rotation files → load into a pygplates.RotationModel object

- Topology feature files → load into a pygplates.FeatureCollection object

- Static polygon files → also load into a pygplates.FeatureCollection object

After collecting the file paths, you will use them to create these pyGPlates objects, which are essential for reconstructing plate motions and geological features through time.

```python
# Create RotationModel object from the rotation files
rotation_model_2019 = pygplates.RotationModel(rotation_filenames)

# Create FeatureCollection for topology features
topology_features_2019 = pygplates.FeatureCollection()
for topology_filename in topology_filenames:
    # Omit files with the string "Inactive" in the filepath
    if "Inactive" not in topology_filename:
        topology_features_2019.add(pygplates.FeatureCollection(topology_filename))
    else:
        topology_filenames.remove(topology_filename)

# Create FeatureCollection for static polygons
static_polygons_2019 = pygplates.FeatureCollection(static_polygon_filename)
```

Once we have our rotation model, topology features and static polygons, we can supply them to the `PlateReconstruction` object to construct the plate motion model. The `PlateReconstruction` object contains methods to reconstruct topology features to a specific geological time!

```python
# Create a PlateReconstruction model with the following inputs:
# - rotation_model: required; defines the rotation files
# - topology_features: optional; defines the topological features files
# - static_polygons: optional; defines the static polygons

model = gplately.PlateReconstruction(rotation_model_2019, topology_features_2019, static_polygons_2019)
```
