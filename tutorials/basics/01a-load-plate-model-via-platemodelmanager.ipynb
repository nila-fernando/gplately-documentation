# Loading Plate Models from Plate Model Manager

The `PlateModelManager` is a tool that helps you download and manage plate tectonic reconstruction models. Think of it as a package manager, but for geological datasets. Instead of manually downloading files and setting paths, you can use this tool to access well-known models with a single line of code. 

For more information on how to use the Plate Model Manager, check out [``Plate Model Manager README``](https://github.com/michaelchin/plate-model-manager/blob/main/README.md).

### Installation

Install via pip:

`pip install plate-model-manager`

or via conda:

`conda install conda-forge::plate-model-manager`

Once it is installed, you’ll be able to import and use `PlateModelManager` in your Python scripts.

### Imports

Now, let's begin by importing our required libraries:

```python
# Ensure you have the required packages installed:
# Run `pip install gplately` if not already installed.
import gplately
from plate_model_manager import PlateModelManager
```

### Loading in a Plate Model

Let's use gplately's `PlateModelManager` to create a plate reconstruction model based on the Zahirovic et al. (2022) global plate model, [“Subduction and carbonate platform interactions”](https://rmets.onlinelibrary.wiley.com/doi/10.1002/gdj3.146).

The `PlateModelManager` downloads the necessary reconstruction files for supported models. 

To list all available models, 
1. Use the command `pmm ls` in your terminal or 
2. Use `print(PlateModelManager().get_available_model_names())` in your python script

To load a specific model's data, pass its exact name (case-sensitive) to the get_model() function. The downloaded files will be saved locally in the directory you specify using the 'data_dir' argument.

```python
# Create a Plate Model Manager
pm_manager = PlateModelManager()

# Load & download model data into the path specified by 'data_dir'
# To load a specific model, you can specify the model name
try:
    zahirovic2022_model = pm_manager.get_model("Zahirovic2022", data_dir="../../data/plate-model-repo")
except Exception as e:
    print(f"Error loading model: {e}")
```

From a `PlateModel` object (like zahirovic2022_model), you can access:

- Rotation files: A set of rotation files - files that end in ".rot"

- Topology files: A set of topology feature files - typically of type ".gpml" or ".gpmlz"

- Static polygon files: A set of static polygons

```python
# Fetch components of the loaded model
rotation_model_2022 = zahirovic2022_model.get_rotation_model()
topology_features_2022 = zahirovic2022_model.get_topologies()
static_polygons_2022 = zahirovic2022_model.get_static_polygons()
```

Once we have our rotation model, topology features and static polygons, we can supply them to the `PlateReconstruction` object to construct the plate motion model. The `PlateReconstruction` object contains methods to reconstruct topology features to a specific geological time!

You can explore the full syntax and use-cases in the [GPlately documentation for PlateReconstruction](https://gplates.github.io/gplately/latest/sphinx/html/generated/gplately.PlateReconstruction.html#gplately.PlateReconstruction).

```python
# Create a PlateReconstruction model with the following inputs:
# - rotation_model: required; defines the rotation files
# - topology_features: optional; defines the topological features files
# - static_polygons: optional; defines the static polygons

model = gplately.PlateReconstruction(rotation_model_2022, topology_features_2022, static_polygons_2022)
```

Now that you have your plate motion model, you can reconstruct and plot ancient coastlines, plate boundaries, subduction zones, mid-ocean ridges, deforming regions, fossil locations, and more; unlocking a dynamic view of Earth's geological past!
