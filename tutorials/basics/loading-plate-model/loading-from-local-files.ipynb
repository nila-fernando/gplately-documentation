To generate the plate reconstruction model, you will need three types of files:

- Rotation files: usually ending with .rot

- Topology feature files: usually ending with .gpml or .gpmlz

- Static polygon files

These files contain different geological data and need to be loaded into specific pyGPlates objects:

- Rotation files → load into a pygplates.RotationModel object

- Topology feature files → load into a pygplates.FeatureCollection object

- Static polygon files → also load into a pygplates.FeatureCollection object

After collecting the file paths, you will use them to create these pyGPlates objects, which are essential for reconstructing plate motions and geological features through time.

```python
# Create RotationModel object from the rotation files
rotation_model_2019 = pygplates.RotationModel(rotation_filenames)

# Create FeatureCollection for topology features
topology_features_2019 = pygplates.FeatureCollection()
for topology_filename in topology_filenames:
    # Omit files with the string "Inactive" in the filepath
    if "Inactive" not in topology_filename:
        topology_features_2019.add(pygplates.FeatureCollection(topology_filename))
    else:
        topology_filenames.remove(topology_filename)

# Create FeatureCollection for static polygons
static_polygons_2019 = pygplates.FeatureCollection(static_polygon_filename)
```

Once we have our rotation model, topology features and static polygons, we can supply them to the `PlateReconstruction` object to construct the plate motion model. The `PlateReconstruction` object contains methods to reconstruct topology features to a specific geological time!

```python
# Create a PlateReconstruction model with the following inputs:
# - rotation_model: required; defines the rotation files
# - topology_features: optional; defines the topological features files
# - static_polygons: optional; defines the static polygons

model = gplately.PlateReconstruction(rotation_model_2019, topology_features_2019, static_polygons_2019)
```

### Reconstruct Geological Features

The plate motion model we created can be used to generate plate reconstructions through geological time. Let's reconstruct subduction zones and mid-ocean ridges to 150 Ma.

```python
time = 150 #Ma
```

```python
# --- Plot ridges and trenches at 150 Ma

subduction_data = model.tessellate_subduction_zones(time)   # Tessellate subduction zones at 150 Ma
ridge_data = model.tessellate_mid_ocean_ridges(time)    # Tessellate mid-ocean ridges at 150 Ma

# Create a new figure (16 x 8 inches in size)
fig = plt.figure(figsize=(16, 8))
ax1 = fig.add_subplot(111)

# Plot subduction zone points as blue dots
ax1.scatter(subduction_data[:, 0],  # longitudes
            subduction_data[:, 1],  # latitudes
            color='blue',
            label='Subduction Zones')

# Plot mid-ocean ridge points as red dots
ax1.scatter(ridge_data[:, 0],
            ridge_data[:, 1],
            color='red',
            label='Mid-Ocean Ridges')

ax1.legend()
plt.show()
```

### Plotting Coastlines

This code segment reconstructs and plots the ancient coastlines at a given time in the past (here, 150 Ma, defined above!), using the plate motion model 'model' that we've created.

- First, we load the present-day coastlines from the model, typically by reading a shapefile and converting it into a FeatureCollection object.

- Next, we reconstruct (move) those coastlines back in time, based on how tectonic plates have shifted.

- Finally, we plot them on a world map using matplotlib as green solid lines.

This is useful for visualizing what Earth’s surface looked like millions of years ago!

```python
# Get present-day coastline features from the model data directory
coastline_filename = f'{plate_model_dir}/StaticGeometries/Coastlines/Global_coastlines_2019_v1_low_res.shp'
coastline_features = pygplates.FeatureCollection(coastline_filename)

# Reconstruct those coastlines to a past geological time (in Ma = million years ago)
reconstructed_coastlines = model.reconstruct(
    coastline_features,
    to_time=time
)

# Set up a plot
fig, ax = plt.subplots(figsize=(12, 8))

# Loop through each reconstructed coastline and plot it
for reconstructed_coastline in reconstructed_coastlines:
    geometry = reconstructed_coastline.get_reconstructed_geometry()
    if geometry:
        coords = geometry.to_lat_lon_array()
        lons = coords[:, 1]  # Longitude
        lats = coords[:, 0]  # Latitude
        ax.plot(lons, lats, 'g-', linewidth=0.5)  # g = green, - = solid line

# Set plot limits, labels & title
ax.set_xlim(-180, 180)
ax.set_ylim(-90, 90)
ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')
ax.set_title(f'Coastlines at {time} Ma')
ax.grid(True)

plt.show()
```
